```python
import pandas as pd
import os 
```

```python
# read the book_ratings.csv file
file_path = "/Users/avikumart/Documents/GitHub/Book-recommendation-system/data/book_ratings.csv"
book_ratings = pd.read_csv(file_path)
```

```python
# sample the radome 100k rows from the book_ratings dataframe
sampled_book_ratings = book_ratings.sample(n=10000, random_state=42)
sampled_book_ratings.info()
```

```python
# save the sampled_book_ratings dataframe to a csv file
output_file_path = "/Users/avikumart/Documents/GitHub/Book-recommendation-system/data/sampled_book_ratings.csv"
sampled_book_ratings.to_csv(output_file_path, index=False)
```

```python
user_item_matrix = sampled_book_ratings.pivot_table(index='user_id', columns='isbn', values='book_rating').fillna(0)
```

```python
user_item_matrix
```

```python
# save the user_item_matrix to a csv file
output_file_path = "/Users/avikumart/Documents/GitHub/Book-recommendation-system/data/user_item_matrix.csv"
user_item_matrix.to_csv(output_file_path, index=True)
```
